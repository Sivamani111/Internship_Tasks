# Q10 Find All Characters Appearing More Than Once

### Step-1 Observe

Input = programming

Expected output = ['r', 'g', 'm']


### Step-2 Understand

### Step-3 Break

```python
1 Check whether the string is empty
2 Create an empty dictionary
3 Count each character
4 Create a list for repeated characters
5 Add characters whose frequency is greater than 1
6 Return the list
```

### Step-4 Logic

### Step-5 Code

```python
def repeated_chars(text):
    frequency = {}

    for char in text:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    repeated = []

    for char in frequency:
        if frequency[char] > 1:
            repeated.append(char)

    return repeated
```

### Step-1 Observe

### Step-2 Understand

### Step-3 Break

```python
1 take sentence as input
2 create an empty dictionary
3 split each word using split()
4 assume first word as most repeated word
5 using for loop iterating the each word 
6 compare the each word with the most reapted word which is first word
7 return most repeated word
```

### Step-4 Logic

### Step-5 Code

```python
def most_frequent_word(Text):
    if Text == "":
        return None
    words = Text.split()
    result = {}
    for word in words:
        if word in result:
            result[word]+=1
        else:
            result[word]=1 
    most_frequent = words[0]
    for word in result:
        if result[word]>result[most_frequent]:
            most_frequent = word
    return most_frequent
```
