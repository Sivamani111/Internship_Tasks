## Q35

```python
import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

# Prediction = [[1,99],[3,4]]
```

```python
#Explanation = 1. a = [[1, 2], [3, 4]] creates a nested list
# 2. b = copy.copy(a) creates a shallow copy of the outer list
# 3. The inner lists are still shared between a and b
# 4. b[0][1] = 99 changes the second element of the first inner list
# 5. Because that inner list is shared, a also reflects the change
```

## Q36

```python
import copy
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)


# Prediction = [[1,2],[3,4]]
```

```python
# Explanation = 1. a = [[1, 2], [3, 4]] creates a two inner lists
# 2. copy.deepcopy(a) creates a completely independent copy, including the inner lists
# 3. b[0][1] = 99 changes the second element of the first inner list in b
# 4. The original list a remains unchanged because its inner lists are independent of b's inner lists
```
